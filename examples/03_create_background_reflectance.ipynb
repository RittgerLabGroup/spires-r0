# Create annual background reflectance in memory

Use the public VIIRS or MODIS builder with an explicit inventory of summer scenes for one tile. All scenes are held in RAM before selection. Request sufficient memory for the stack and selection intermediates (several hundred GiB for full tiles). Products are written atomically as NetCDF.


```python
from pathlib import Path
from spires_r0 import build_viirs_r0_from_sources, build_modis_r0_from_sources
```

## VIIRS

Replace these example filenames with the complete, date-ordered summer inventory for your tile. The builder reads native scenes, applies the annual selector, and validates the saved R0.


```python
viirs_sources = [
    Path("VJ109GA.A2025152.h09v05.002.2025200120000.h5"),
    Path("VJ109GA.A2025182.h09v05.002.2025200120000.h5"),
]
viirs_r0 = build_viirs_r0_from_sources(
    viirs_sources, r0_path="viirs_r0.nc", show_progress=True,
)
viirs_r0.close()
```

## MODIS

The MODIS builder uses the same in-memory execution method with its sensor-specific reader and bands.


```python
modis_sources = [
    Path("MOD09GA.A2025152.h09v05.061.2025200120000.hdf"),
    Path("MOD09GA.A2025182.h09v05.061.2025200120000.hdf"),
]
modis_r0 = build_modis_r0_from_sources(
    modis_sources, r0_path="modis_r0.nc", show_progress=True,
)
modis_r0.close()
```
